```python
# So after some research I get to know : 
'''
Normalization is applied when we dont know the distribution of the dataset and typically when we want the values to be in between a specific
range (usually 0 and 1).
Standardization is applied so that we have dataset having gaussian distribution , we do this step to get : 
standard_deviation = 1, mean = 0

'''
```

```python
import numpy as np
from sklearn.preprocessing import StandardScaler,MinMaxScaler
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
X = np.random.rand(10,1).reshape(-1,1)
```

```python
X
```

```python
StandardScaler().fit_transform(X)
```

```python
MinMaxScaler().fit_transform(X)
```

```python
plt.figure(figsize=(15,5))

plt.subplot(131)
sns.kdeplot(data=X)
plt.title('Normal Data')

plt.subplot(132)
sns.kdeplot(data=StandardScaler().fit_transform(X))
plt.title('Data after Standardization')

plt.subplot(133)
sns.kdeplot(data=MinMaxScaler().fit_transform(X))
plt.title('Data after Normalizaion')

plt.show()
```

```python
# NOTE : The shape haven't changed but note the ranges in the bottom
# Standardization have shifted the mean to be 0 now and the most of variance can be explained by +-3td 
# Normalization graph shows how the values were squished to in between 0 and 1 reflecting in y axis
```

    Feature	        Standardization	                                    Normalization
    ______________________________________________________________________________________________________________________
    Range	        Unbounded(usually -3 to 3)	                    Strictly 0 to 1 (or -1 to 1)
    Mean	        Always 0	                                    Varies
    Outliers	Less affected(keeps their relative distance)	    Highly affected(outliers compress the rest of the data)
