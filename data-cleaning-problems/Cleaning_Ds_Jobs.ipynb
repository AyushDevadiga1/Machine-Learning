```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('datasets\Uncleaned_DS_jobs.csv',index_col='index')
```

```python
df.shape
```

```python
df.columns
```

```python
# Header cleaning
df.columns = df.columns.str.lower().str.replace(' ','_')
```

```python
df.head()
```

```python
df.tail()
```

```python
# Notice that there is -1 in many rows this is a sign that data is missing there .
```

```python
df.isna().sum() # This is just a scam for now as we know that data is having -1 as na
```

```python
# So to actually get the nan we will do some replacements.
# Twice cause I want to make sure both string and numeric datatype is handled.
# Note : There is no data that might have -1 as valid data here so replacing is fine .

df = df.replace(-1,np.nan)
df = df.replace('-1',np.nan)
```

```python
df.isna().sum()/len(df)*100 
# Now this is the reality, or who knows something else may be lurking underneath ; If we will find it later.
```

```python
# For further steps I use data wrangler for EDA
```

```python
df['salary_estimate'].str.split(" ")[0][0]
```

```python
df['salary_estimate'] = df['salary_estimate'].str.split(" ").str[0] # Some Standard Cleaning for Salary Estimate
```

```python
df.sample()
```

```python
# There is strangely the rating again specified in the company name so we are gonna remove it . 
df['company_name'].str.split('\n').str[0]
```

```python
df['company_name'] = df['company_name'].str.split('\n').str[0]
```

```python
df.sample(3)
```

```python
# Found another thing now that the revenue column have rows with value : Unknown / Non-Applicable which are also nan so we must handle this too.
```

```python
df['revenue'] = df['revenue'].replace('Unknown / Non-Applicable',np.nan)
```

```python
df.isna().sum()/len(df)*100  
# As we can see the Revenue null values have spiked .
```
