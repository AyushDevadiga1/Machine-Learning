```python
import matplotlib.gridspec as gridpsec
```

```python
fig = plt.figure(figsize=(15,10))

gsp = gridpsec.GridSpec(
                            nrows = 3,
                            ncols = 2,
                            height_ratios=[1,1,1],
                            width_ratios=[1,1],
                            figure = fig
)

ax1 = fig.add_subplot(gsp[0:1,0:1])
ax2 = fig.add_subplot(gsp[0:1,1:2])
ax3 = fig.add_subplot(gsp[1:2,0:1])
ax4 = fig.add_subplot(gsp[1:2,1:2])
ax5 = fig.add_subplot(gsp[2:,0:1])
ax6 = fig.add_subplot(gsp[2:,1:2])

colors = sns.color_palette("rocket", 7)

ax1.hist(df['Confirmed'],color=colors[1])
ax2.plot(df['Confirmed'],color=colors[2])
ax3.hist(df['Deaths'],color=colors[3])
ax4.plot(df['Deaths'],color=colors[4])
ax5.hist(df['Recovered'],color=colors[5])
ax6.plot(df['Recovered'],color=colors[6])


plt.tight_layout()
plt.show()
```
