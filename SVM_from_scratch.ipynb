## Assumptions for SVM classifier : 

Key assumptions and requirements for SVM classification include:

    Linear Separability or High-Dimensional Mapping: The fundamental assumption is that data is linearly separable,
    or can be projected into a higher-dimensional space where it becomes linearly separable using kernels. 

    Maximum Margin Optimization: SVM assumes that the best separation between classes is achieved by 
    maximizing the distance (margin) between the decision boundary and the closest data points from either class. 

    Support Vectors Define the Boundary: The model assumes that only the points closest to the hyperplane (support vectors)
    are necessary for determining the classification boundary; other points are irrelevant, as noted on Brilliant.org and by Medium users. 

    Feature Scaling Requirements: SVM assumes that all input features are on a similar scale.
    Since SVM calculates distances, unscaled data will cause features with larger magnitudes to dominate the model,
    requiring normalization or standardization. 

    Data Noise Tolerance: While Hard Margin SVM assumes perfectly separated data, 
    Soft Margin SVM relaxes this, assuming that allowing some misclassifications or violations of the margin helps create a better-generalizing model. 

    Independence of Observations: Like many machine learning models, SVM assumes that data samples are independent and identically distributed.
