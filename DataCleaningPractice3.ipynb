```python
df = pd.read_csv('wiki_messy.csv')
```

```python
df.columns
```

```python
df.columns = df.columns.str.lower().str.replace(' ',"_").str.replace(r'\xa0','_',regex=True).str.replace('.',"")
```

```python
df.info()
```

```python
print(df['peak'].str.replace(r'\d','X',regex=True).value_counts())
print('-'*145)
print(df['all_time_peak'].str.replace(r'\d','X',regex=True).value_counts())
print('-'*145)
print(df['ref'].str.replace(r'\d','X',regex=True).value_counts())
print('-'*145)
```

```python
print(df['actual_gross'].str.replace(r'\d','x',regex=True).value_counts())
print('-'*145)
print(df['adjusted_gross_(in_2022_dollars)'].str.replace(r'\d','X',regex=True).value_counts())
print('-'*145)
print(df['average_gross'].str.replace(r'\d','X',regex=True).value_counts())
```

```python
df['actual_gross'] = df['actual_gross'].str.findall(r'[0-9]').str.join('').astype('int')
```

```python
df['adjusted_gross_(in_2022_dollars)'] = df['adjusted_gross_(in_2022_dollars)'].str.findall(r'[0-9]').str.join('').astype('int')
```

```python
df['average_gross'] = df['average_gross'].str.findall(r'[0-9]').str.join('').astype('int')
```

```python
#------------------------------------------------------------------------------------------------------------------------------------------#
```

```python
df['peak'] = pd.to_numeric(df['peak'].str.extract(r'^([^\[]+)')[0],errors='coerce').astype('Int64')
```

```python
df['all_time_peak'] = pd.to_numeric(df['all_time_peak'].str.extract(r'^([^\[]+)')[0],errors='coerce').astype('Int64')
```

```python
df['year_start'] = df['year(s)'].str.split('–').str[0]
df['year_end'] = df['year(s)'].str.split('–').str[1]
```
