```python
from sklearn.model_selection import train_test_split,cross_val_score    
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression,Ridge
from sklearn.metrics import r2_score,root_mean_squared_error

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv('datasets\diabetes.csv')
```

```python
df.head()
```

```python
X,y = df.drop(columns=['Outcome']),df['Outcome']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2) # Splitting the dataset
```

```python
alphas = [0.1 , 1 , 10 , 15 , 25, 50] # Test values of alpha
weights = []

for alpha in alphas:
    las = Ridge(alpha=alpha) # Instantiation with current alpha value
    las.fit(X_train,y_train) # Fit the training set 
    y_pred_las = las.predict(X_test) # Predict using Lasso model 
    weights.append(las.coef_)
weights_df = pd.DataFrame(weights,columns=df.drop(columns=['Outcome']).columns,index=alphas)
```

```python
weights_df # This is the weights and how they are updated as the value of alpha increases
```

```python
# Plotting each coefficient as a segment of a bar for each Alpha
weights_df.plot(kind='bar', stacked=False, figsize=(12, 6), edgecolor='black')

plt.axhline(0, color='black', linewidth=1)
plt.title('Feature Coefficients across different Alpha values')
plt.xlabel('Alpha')
plt.ylabel('Coefficient Value')
plt.legend(title='Features', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
# Plotting the transpose of the dataframe so each column (feature) is a line
for column in weights_df.columns:
    plt.plot(weights_df.index, weights_df[column], marker='o', label=column)

plt.xscale('log') # Alpha spans different magnitudes; log scale makes it readable
plt.axhline(0, color='black', linestyle='--', linewidth=1, alpha=0.5)
plt.xlabel('Alpha (Regularization Strength)')
plt.ylabel('Coefficient Value')
plt.title('Ridge Coefficient Paths:')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True, which="both", ls="-", alpha=0.2)
plt.show()
```
